```python
import json
kafka_connection_json=dbutils.secrets.get(scope="finguard-scope",key="kafka_connection_details")
kafka_config=json.loads(kafka_connection_json)
bootstrap_servers=kafka_config['bootstrap_servers']
api_key=kafka_config['api_key']
api_secret=kafka_config['api_secret']
topic=kafka_config['topic']
```

```python
print(f"topic is {topic}")
```

```python
jaas_config=f'kafkashaded.org.apache.kafka.common.security.plain.PlainLoginModule required username="{api_key}" password="{api_secret}";'
```

```python
sample_batch=(spark.read.format("kafka")
              .option("kafka.bootstrap.servers",bootstrap_servers)
              .option("subscribe",topic)
            .option("kafka.security.protocol", "SASL_SSL")
        .option("kafka.sasl.mechanism", "PLAIN")
        .option("kafka.sasl.jaas.config", jaas_config)
        .option("startingOffsets","earliest")
        .load()
)
```

```python
streaming_df=(spark.readStream.format("kafka")
              .option("kafka.bootstrap.servers",bootstrap_servers)
              .option("subscribe",topic)
            .option("kafka.security.protocol", "SASL_SSL")
        .option("kafka.sasl.mechanism", "PLAIN")
        .option("kafka.sasl.jaas.config", jaas_config)
        .option("startingOffsets","earliest")
        .load()
)
```

```python
sample_batch.count()
```

```python

display(sample_batch)
```

```python
from pyspark.sql.functions import col
parsed_batch=sample_batch.select(
col("key").cast("string"),
col("value").cast("string"),
col("topic"),
col("partition"),
col("offset"),
col("timestamp"),
col("timestampType")
)
```

```python
display(parsed_batch)
```

```python
create_catalog = spark.sql(f"CREATE CATALOG IF NOT EXISTS finguard")
create_schema = spark.sql(f"CREATE SCHEMA IF NOT EXISTS finguard.bronze")
```

```python
parsed_batch.write.saveAsTable("finguard.bronze.transactions_batch_test")
```

```python
%sql

select * from finguard.bronze.transactions_batch_test limit 10
```

```python
from pyspark.sql.functions import col
parsed_streaming_df=streaming_df.select(
col("key").cast("string"),
col("value").cast("string"),
col("topic"),
col("partition"),
col("offset"),
col("timestamp"),
col("timestampType")
)
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS finguard.source;
CREATE VOLUME IF NOT EXISTS finguard.source.transactions;
```

```python
streaming_query=(parsed_streaming_df.writeStream.format("delta")
                .outputMode("append")
                .option("checkpointLocation", "/Volumes/finguard/source/transactions/checkpoint/")
                .trigger(availableNow=True)
                .toTable("finguard.bronze.transactions_streaming_test")
)
```

```python
print("Query Started with query id: ",streaming_query.id)
```

```python
%sql

select count(*) from finguard.bronze.transactions_streaming_test
```
